## Recherche des attributs préférés

La distribution des 100 points se fait parmis els attributs suivants:  


| Attribut        | Colonne  |
|-----------------|----------|
| Attractive      | attr1_1  |
| Sincere         | sinc1_1  |
| Intelligent     | intel1_1 |
| Fun             | fun1_1   |
| Ambitious       | amb1_1   |
| Shared interest | shar1_1  |

Ce systeme de notation est relatif: pour qu'une categorie gagne un point, une autre doit en perdre un. 

Les vagues 6 à 9 doivent etre excluent car elles utilisent un systeme de notation different, systeme de notation sur 10. Ce systeme de notation est absolu.

La notation est declarative: elle se base sur ce que les participants disent de leur preferences.

```python
attributs = ["attr1_1", "sinc1_1", "intel1_1", "fun1_1", "amb1_1", "shar1_1"]  # attributs declaratif de preferences du participant

# exclusion des vagues 6 à 9
df_prefs_rel = df_participants[~df_participants["wave"].isin([6, 7, 8, 9])]

df_prefs_rel.groupby("gender")[attributs].mean().round(1)
```

### Diagramme en barres

```python
df_moyennes = df_prefs_rel.groupby("gender")[attributs].mean().round(1).reset_index()
df_moyennes["gender"] = df_moyennes["gender"].map({0: "Femmes", 1: "Hommes"})

df_long = df_moyennes.melt(id_vars="gender", var_name="attribut", value_name="points")

fig = px.bar(df_long, x="attribut", y="points", color="gender", barmode="group",
             title="Importance déclarée des six attributs, par sexe (hors vagues 6-9)",
             labels={"points": "Points alloués sur 100", "attribut": "", "gender": ""})
fig.show()
```



Les facteurs predominants sont 
- attractivité
- sincétité
- intelligence
- humour

L'importance pour l'attribut d'intelligence domine chez les femmes.  
L'importance pour l'attribut d'attractivité domine chez les hommes. L'ecart semble significatif.    
Les hommes semblent attribuer une importance moindre à l'ambition relativement au femmes.

```python
hommes = df_prefs_rel[df_prefs_rel["gender"] == 1]["attr1_1"].dropna()
femmes = df_prefs_rel[df_prefs_rel["gender"] == 0]["attr1_1"].dropna()

res = stats.ttest_ind(hommes, femmes, equal_var=False)

print(f"Moyenne hommes : {hommes.mean():.1f}")
print(f"Moyenne femmes : {femmes.mean():.1f}")
print(f"Probabilité que cet écart soit dû au hasard : {res.pvalue:.10f}")
```

```python
seuil = 0.05
verdict = "rejetée" if res.pvalue < seuil else "non rejetée"
print(f"H0 (les deux moyennes sont égales) {verdict} au seuil de {seuil:.0%}")
```

Les hommes accordent en moyenne 29,4 points sur 100 à l'attractivité, contre 18,8 chez les femmes. La probabilité que cet écart provienne du seul hasard de l'échantillon est quasi nulle : l'hypothèse d'égalité des moyennes est rejetée au seuil de 5 %. L'attractivité est le principal point de divergence entre hommes et femmes dans les préférences déclarées.

## Validité de l'auto-appréciation

```python
soi   = ["attr3_1", "sinc3_1", "intel3_1", "fun3_1", "amb3_1"]
recus = ["attr_recu", "sinc_recu", "intel_recu", "fun_recu", "amb_recu"]

moyennes_recues = (df.groupby("iid")[["attr_o", "sinc_o", "intel_o", "fun_o", "amb_o"]]
                     .mean().set_axis(recus, axis=1))
df_auto = df_participants.merge(moyennes_recues, on="iid")

bilan = pd.DataFrame({
    "attribut": ["Attractivité", "Sincérité", "Intelligence", "Humour", "Ambition"],
    "auto-évaluation": df_auto[soi].mean().values,
    "note reçue": df_auto[recus].mean().values,
    "corrélation": [df_auto[[s, r]].corr().iloc[0, 1] for s, r in zip(soi, recus)],
}).round(2)
bilan.insert(3, "écart", (bilan["auto-évaluation"] - bilan["note reçue"]).round(2))
bilan
```

```python
fig = px.bar(bilan.melt(id_vars="attribut", value_vars=["auto-évaluation", "note reçue"],
                        var_name="source", value_name="note"),
             x="attribut", y="note", color="source", barmode="group",
             title="Ce que les participants pensent d'eux-mêmes vs ce que leurs partenaires leur attribuent",
             labels={"note": "Note moyenne (1-10)", "attribut": "", "source": ""})
fig.show()
```

Sur les cinq attributs, la corrélation entre auto-évaluation et note reçue est faible, comprise entre −0,01 et 0,29. Même sur l'attractivité, l'attribut le mieux prédit, l'auto-évaluation ne capte que 8,4 % de la variance de la note reçue.

# 5. Conclusion

La problématique était de savoir si ce que les participants déclarent rechercher correspond à ce qui détermine réellement leur décision. La réponse est non.

Avant la soirée, les participants répartissent leurs 100 points de façon équilibrée : les femmes placent l'intelligence en tête (21,5 points), puis l'attractivité (18,8) et la sincérité (18,2) ; les hommes placent l'attractivité nettement devant (29,4), un écart avec les femmes que le test de Student confirme comme significatif. Dans les deux cas, l'ambition et les intérêts communs ferment la marche.

Après la rencontre, une seule note explique vraiment la décision : l'attractivité, corrélée à 0,49 avec le « oui », et ce pour les deux sexes. Le taux de « oui » passe de 6 % pour une note de 3 à 85 % pour une note de 10. L'humour (0,41) et le sentiment d'intérêts partagés (0,40) viennent ensuite. La sincérité et l'intelligence, pourtant mises en avant dans les déclarations, ne pèsent que 0,21 et 0,22.

Les autres pistes sont marginales : l'origine commune n'ajoute que deux points de « oui », la corrélation objective des hobbies ne joue pas, et l'ordre de passage n'a pas d'effet au-delà d'un léger avantage pour le tout premier rendez-vous. Enfin, les participants se surévaluent sur tous les attributs d'environ un point sur dix, et leur auto-évaluation prédit mal la note que leur partenaire leur donne.

Pour Tinder, la leçon est que le discours des utilisateurs sur leurs critères est un mauvais guide : ce qui déclenche un match tient à l'attractivité perçue et à une impression d'affinité immédiate, bien plus qu'aux qualités que les profils mettent en avant.
